```python
import pandas as pd

df = pd.read_csv("/content/aotizhongxin_features.csv")

df["datetime"] = pd.to_datetime(
    df[["year", "month", "day", "hour"]]
)

df = df.set_index("datetime")

print("Feature dataset loaded successfully:", df.shape)
display(df.head())
```

```python
# Seperate x and y
TARGET = "PM2.5"

X = df.drop(columns=[TARGET])
y = df[TARGET]

print("Feature matrix shape:", X.shape)
print("Target vector shape:", y.shape)
```

```python
# Perform time based split

n = len(df)

train_end = int(n * 0.70)
val_end = int(n * 0.85)

X_train = X.iloc[:train_end]
y_train = y.iloc[:train_end]

X_val = X.iloc[train_end:val_end]
y_val = y.iloc[train_end:val_end]

X_test = X.iloc[val_end:]
y_test = y.iloc[val_end:]
```

```python
# Verify split sizes
print("Train set:", X_train.shape, y_train.shape)
print("Validation set:", X_val.shape, y_val.shape)
print("Test set:", X_test.shape, y_test.shape)
```

```python
# Time range check
print("Train period:", X_train.index.min(), "→", X_train.index.max())
print("Validation period:", X_val.index.min(), "→", X_val.index.max())
print("Test period:", X_test.index.min(), "→", X_test.index.max())
```

```python
from pathlib import Path

OUTPUT_DIR = Path("/content/beijing_data/processed")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

X_train.to_csv(OUTPUT_DIR / "X_train.csv", index=False)
y_train.to_csv(OUTPUT_DIR / "y_train.csv", index=False)

X_val.to_csv(OUTPUT_DIR / "X_val.csv", index=False)
y_val.to_csv(OUTPUT_DIR / "y_val.csv", index=False)

X_test.to_csv(OUTPUT_DIR / "X_test.csv", index=False)
y_test.to_csv(OUTPUT_DIR / "y_test.csv", index=False)

print("All train, validation, and test files saved successfully!")
```
